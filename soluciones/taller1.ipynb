```python
municipios = (
    ds_agrosavia["Municipio"]
    .dropna()
    .str.strip()
    .loc[lambda s: s.ne("")]
    .drop_duplicates()
    .sort_values()
    .reset_index(drop=True)
)
```

```python
# Nombres de los campos que deben ser numéricos.
campos_numericos = {
    "Secuencial",
    "pH agua:suelo",
    "Materia organica",
    "Fósforo Bray II",
    "Azufre Fosfato monocalcico",
    "Calcio intercambiable",
    "Magnesio intercambiable",
    "Sodio intercambiable",
    "capacidad de intercambio cationico",
}

# Localizar columnas ignorando espacios externos y mayúsculas.
buscados = {nombre.strip().casefold() for nombre in campos_numericos}
columnas = [
    col for col in ds_agrosavia.columns
    if col.strip().casefold() in buscados
]

encontrados = {col.strip().casefold() for col in columnas}
if faltantes := buscados - encontrados:
    raise ValueError(f"Columnas no encontradas: {sorted(faltantes)}")

df_agrosavia_corregido = ds_agrosavia.copy()
incidencias = []

for columna in columnas:
    original = ds_agrosavia[columna]
    texto = original.astype("string").str.strip()

    es_vacio = texto.isna() | texto.eq("")
    numero = pd.to_numeric(texto, errors="coerce")
    es_error = numero.isna() & ~es_vacio

    # Guardar ubicación y contenido original de cada incidencia.
    for mascara, motivo in [
        (es_error, "No convertible a número"),
        (es_vacio, "Valor faltante"),
    ]:
        incidencias.append(pd.DataFrame({
            "indice_fila": ds_agrosavia.index[mascara],
            "columna": columna,
            "valor_original": original.loc[mascara].to_numpy(),
            "motivo": motivo,
        }))

    df_agrosavia_corregido[columna] = numero

df_errores = pd.concat(incidencias, ignore_index=True)

# Resumen por columna y motivo.
display(
    df_errores.groupby(["columna", "motivo"])
    .size()
    .rename("cantidad")
    .reset_index()
)

# Detalle: índice, columna y valor que requiere revisión.
display(df_errores)
```
