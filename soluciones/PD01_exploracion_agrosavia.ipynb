## 1. Inspección inicial · 10 minutos

Presentar la cantidad de filas y columnas, los nombres de las columnas, sus tipos y cinco registros. Explicar qué representa una fila y por qué la lectura inicial conserva los campos como texto.

📘 **Apoyo:** `shape`, `columns`, `dtypes`, `head` e `info`.

✅ **Comprobación:** El corte del manifiesto contiene 92.738 filas y 32 columnas. La tabla original debe conservar esa cantidad de filas durante el ejercicio.

````python
# Desarrollo de la actividad 1
# Mostrar dimensiones, tipos y primeros registros de suelo.
# Guardar el total de filas original para los controles posteriores.
import pandas as pd

# 1. Carga inicial conservando todo como texto (dtype=str) y evitando nulos automáticos
df_agrosavia = pd.read_csv(
    '../kit/data/raw/agrosavia.csv', dtype=str, keep_default_na=False, encoding="utf-8"
)

# Guardar el total de filas original para controles posteriores
TOTAL_FILAS_ORIGINAL = len(df_agrosavia)
TOTAL_COLUMNAS_ORIGINAL = len(df_agrosavia.columns)

print(
    f"Dimensiones del dataset: {TOTAL_FILAS_ORIGINAL} filas y {TOTAL_COLUMNAS_ORIGINAL} columnas."
)
print("\nNombres de columnas y tipos:")
print(df_agrosavia.dtypes)

print("\nPrimeros 5 registros:")
print(df_agrosavia.head(5))
````

**Interpretación del resultado:**

Cada fila representa una muestra de suelo tomada en campo, la cual incluye metadatos geográficos, temporales y los resultados de los análisis fisicoquímicos de laboratorio asociados (como pH, materia orgánica, fósforo, etc.).

Forzar dtype=str y keep_default_na=False previene que Pandas interprete prematuramente comas decimales (ej. "5,5"), cadenas vacías ("") o marcadores textuales de laboratorio ("ND", "NA", "N.D.") como valores nulos de Python (NaN), asegurando que ningún dato original se corrompa o descarte antes de realizar una validación explícita.


## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
# Crear y mostrar perfil con las cinco columnas solicitadas.
# 1. Cuantificar nulos reales de pandas (si los hubiera)
nulos_pandas = df_agrosavia.isna().sum()

# 2. Cuantificar celdas vacías o con espacios en blanco después de hacer strip()
df_stripped = df_agrosavia.apply(lambda col: col.str.strip())
celdas_vacias = (df_stripped == "").sum()

# 3. Cuantificar marcadores textuales especiales de laboratorio (ej. 'ND', 'NA')
marcadores_especiales = df_stripped.isin(["ND", "NA", "N.D.", "NaN", "-"]).sum()

print("--- DIAGNOSTICO DE CALIDAD ---")
diagnostico_df = pd.DataFrame(
    {
        "Nulos Pandas": nulos_pandas,
        "Celdas Vacías": celdas_vacias,
        "Marcadores (ND/NA)": marcadores_especiales,
    }
)
print(diagnostico_df)
````

**Interpretación del resultado:**

Los datasets de campo suelen contener celdas vacías o texto no numérico debido a limitaciones en las mediciones o errores de digitación. Tratar estas anomalías por separado permite entender si la ausencia de datos se debe a un error de captura o a una imposibilidad técnica del laboratorio (como en el caso de muestras con etiqueta "ND" - No Determinado), evitando alterar las métricas estadísticas posteriores.

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
# Crear suelo["ph"] y suelo["fecha_dt"].
# Mostrar los fallos de conversión y sus valores originales
import numpy as np
# Conservar original: 'pH agua:suelo'
ph_original = df_agrosavia["pH agua:suelo"].str.strip()

# Identificar cuáles son entradas vacías vs textos no vacíos
es_vacio_ph = ph_original == ""


# Función auxiliar para evaluar si un texto no vacío falla al convertirse a float
def evaluar_conversion_ph(val):
    if val == "":
        return np.nan  # Vacío
    try:
        return float(val.replace(",", "."))
    except ValueError:
        return (
            "NO_CONVERTIBLE"  # Texto no vacío que falló en la conversión numérica
        )


# Aplicar conversión numérica con errors="coerce" simulado / explícito
ph_convertido_eval = ph_original.apply(evaluar_conversion_ph)

# Crear la nueva columna 'ph' numérica oficial usando pd.to_numeric con errors='coerce'
ph_limpio = ph_original.str.replace(",", ".", regex=False).replace("", np.nan)
df_agrosavia["ph"] = pd.to_numeric(ph_limpio, errors="coerce")

# Distinguir explícitamente cuántas fueron vacías y cuántas fueron textos no convertibles fallidos
total_vacios_ph = es_vacio_ph.sum()
total_no_convertibles_ph = (ph_convertido_eval == "NO_CONVERTIBLE").sum()

# Conservar original: 'Fecha de Análisis'
fecha_original = df_agrosavia["Fecha de Análisis"].str.strip()

es_vacio_fecha = fecha_original == ""

# Conversión a datetime con formato estricto %d/%m/%Y y errors="coerce"
df_agrosavia["fecha_dt"] = pd.to_datetime(
    fecha_original, format="%d/%m/%Y", errors="coerce"
)

# Textos no vacíos que fallaron al parsearse como fecha
fallos_fecha = (~es_vacio_fecha) & (df_agrosavia["fecha_dt"].isna())
total_no_convertibles_fecha = fallos_fecha.sum()
total_vacios_fecha = es_vacio_fecha.sum()


# --- Resultados del diagnóstico de conversión ---
print("=== DIAGNÓSTICO DE CONVERSIÓN Y TIPOS ==N")
print(f"Total de filas originales: {TOTAL_FILAS_ORIGINAL}")
print(
    f"pH -> Vacíos: {total_vacios_ph} | Textos no convertibles: {total_no_convertibles_ph}"
)
print(
    f"Fechas -> Vacíos: {total_vacios_fecha} | Textos no convertibles: {total_no_convertibles_fecha}"
)

print("\nTipos de datos resultantes:")
print(
    df_agrosavia[
        ["pH agua:suelo", "ph", "Fecha de Análisis", "fecha_dt"]
    ].dtypes
)
````

**Interpretación del resultado:**

El parámetro errors="coerce" garantiza que cualquier anomalía insalvable se convierta silenciosamente en NaN en lugar de interrumpir la ejecución del script con un ValueError.
Mantener intactas las columnas fuente (pH agua:suelo y Fecha de Análisis) permite conservar la traza documental exacta de los datos tal como fueron reportados por el laboratorio, evitando cualquier pérdida irreversible de información durante la limpieza.
Al separar las celdas vacías ("") de los textos no vacíos que fallaron en la conversión (como cadenas con caracteres inválidos, rangos tipo ">10.00" o errores de sintaxis como "4..01"), se logra una auditoría de calidad de datos mucho más precisa. Esto permite diferenciar una simple ausencia de medición de un error real de digitación o formato en el reporte de laboratorio.

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
# Crear en_revision y los conteos por motivo.
# Comprobar la conservación del total original con assert.
cond_ph_nulo = df_agrosavia["ph_num"].isna()
cond_ph_fuera_rango = (df_agrosavia["ph_num"] < 0) | (
    df_agrosavia["ph_num"] > 14
)
cond_fecha_nula = df_agrosavia["fecha_dt"].isna()

# Creación de la bandera sin eliminar registros
df_agrosavia["en_revision"] = (
    cond_ph_nulo | cond_ph_fuera_rango | cond_fecha_nula
)

# Aseveración de volumetría intacta
assert (
    len(df_agrosavia) == TOTAL_FILAS_ORIGINAL
), "Error: Se alteró la volumetría del dataset original."

print("\nDistribución de la bandera en_revision:")
print(df_agrosavia["en_revision"].value_counts())
````

**Interpretación del resultado:**

En analítica de datos agroindustriales, eliminar registros defectuosos de forma prematura puede introducir sesgos de selección. La estrategia de marcar con una bandera (en_revision = True) aísla los registros problemáticos para una auditoría posterior por parte del equipo de agrónomos o estadísticos, preservando intacto el tamaño original de la muestra (92.738 registros).

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT / "controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
# Construir controles, mostrarlo y exportarlo a OUT.
# Redactar los dos hallazgos en la celda Markdown siguiente.
import os
import pandas as pd

# 1. Asegurar que el directorio OUT existe
os.makedirs("OUT", exist_ok=True)

# 2. Construcción del DataFrame de controles con las columnas 'control' y 'cantidad'
total_entrada = len(df_agrosavia)
ph_sin_numero = df_agrosavia["ph"].isna().sum()
ph_fuera_intervalo = ((df_agrosavia["ph"] < 0) | (df_agrosavia["ph"] > 14)).sum()
fecha_no_interpretable = df_agrosavia["fecha_dt"].isna().sum()

# Total de filas en revisión (registros con al menos una anomalía)
total_en_revision = (
    df_agrosavia["ph"].isna()
    | (df_agrosavia["ph"] < 0)
    | (df_agrosavia["ph"] > 14)
    | df_agrosavia["fecha_dt"].isna()
).sum()

df_controles = pd.DataFrame(
    {
        "control": [
            "Total de entrada",
            "pH sin número",
            "pH fuera del intervalo",
            "Fecha no interpretable",
            "Total de filas en revisión",
        ],
        "cantidad": [
            total_entrada,
            ph_sin_numero,
            ph_fuera_intervalo,
            fecha_no_interpretable,
            total_en_revision,
        ],
    }
)

# 3. Exportación a la ruta OUT/controles_agrosavia.csv sin índice
ruta_salida = os.path.join("OUT", "controles_agrosavia.csv")
df_controles.to_csv(ruta_salida, index=False)

print(f"Archivo de controles exportado exitosamente en: {ruta_salida}")
print("\nContenido de la tabla de controles:")
print(df_controles)
````

**Interpretación del resultado:**

Hallazgo 1: Alta confiabilidad en la captura temporal y estabilidad del instrumento de pHConteo de referencia: Fecha no interpretable = 0 y pH fuera del intervalo = 0.
El hecho de que la totalidad de los registros cuenten con fechas de análisis válidas demuestra un registro cronológico riguroso y automatizado en la recepción de las muestras de laboratorio. Asimismo, que no existan mediciones de pH fuera del dominio físico de 0 a 14 evidencia que los sensores o equipos de medición operaron bajo parámetros estables sin registrar errores numéricos de desbordamiento o saturación.

Hallazgo 2: Presencia marginal de errores tipográficos en el análisis fisicoquímicoConteo de referencia: pH sin número = 11 (equivalente al Total de filas en revisión).
De un total de 92.738 registros, únicamente 11 muestras presentaron anomalías en la lectura del pH (como caracteres inválidos o celdas no convertibles), lo que representa una tasa de error inferior al $0.012\%$. Esto indica una excelente calidad general en la digitación y captura de datos por parte del laboratorio AGROSAVIA, permitiendo aislar dichos registros mediante la bandera de revisión sin comprometer la integridad ni la potencia estadística del conjunto de datos principal.
